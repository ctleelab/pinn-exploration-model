#### Import and Preprocess Data

```python
import mrcfile
from data_generation.preprocess import keep_if_enough_neighbors, clahe_volume_slicewise_fast, overlay_points_on_cryoet
from skimage.morphology import remove_small_objects

#### IMPORT MRC FILE #####
my_data_path = "../data/experimental/vol/formatted/"
file_path = f"{my_data_path}/{data_id}/{shape}.mrc"
with mrcfile.open(file_path, permissive=True) as mrc:
    cryoet = mrc.data

# #### Preprocess for Golgi ####
# clahe_vol = clahe_volume_slicewise_fast(cryoet, clipLimit=2.0, tileGridSize=(8, 8))
# filtered = (clahe_vol > 0.7).astype(np.uint8)
# filtered = keep_if_enough_neighbors(filtered, radius=2, k=55)
# processed = remove_small_objects(filtered.astype(bool), min_size=60000, connectivity=1)

#### Preprocess for Mitochondria ####
clahe_vol = clahe_volume_slicewise_fast(cryoet, clipLimit=2.0, tileGridSize=(8, 8))
filtered  = (clahe_vol > 0.6).astype(np.uint8)
processed = keep_if_enough_neighbors(filtered, radius=2, k=55)

#### Visualize Data ####
visualize = processed
overlay_points_on_cryoet(cryoet=visualize, axis='x', slice_index=slices, show_title=True)
```

#### Edge/Bulk Signal

```python
from data_generation.preprocess import downsample_binary_mask_to_fraction, overlay_points_on_cryoet, build_data_edge_from_masks
from pinn.utils import save_pts_data, load_pts_data

# #### Parameters for Golgi  ####
# threshold = 0.8
# target_edge = 1.
# target_bulk = 0.05
# mask_remove = None

#### Parameters for Mitochondria  ####
threshold = 0.8
target_edge = 0.004
target_bulk = 0.01
Z = processed.shape[0]
z = np.arange(Z)[:, None, None]   # shape (Z,1,1)
mask_remove = ((z >= 154) & (z <= 164)) | ((z >= 320) & (z <= 330))

#### Sample edge/bulk signals ####
edge = (processed > threshold)
bulk = (~edge)
if mask_remove is not None:
    bulk = bulk & (~mask_remove)
edge = downsample_binary_mask_to_fraction(edge, target_frac=target_edge, seed=0)
bulk = downsample_binary_mask_to_fraction(bulk, target_frac=target_bulk, seed=0)
edge = edge.astype(np.float32)
bulk = bulk.astype(np.float32)
print("edge size: ", edge.sum())
print("bulk size: ", bulk.sum())

#### Save point data ####
data_edge = build_data_edge_from_masks(edge=edge, bulk=bulk, volume_shape=clahe_vol.shape)
meta_data ={
    "type": "edge",
    "label_meaning": {1: "edge", 0: "bulk"},
    "GRID_SIZE": clahe_vol.shape[0],
    "edge_min": threshold,
    "target_edge": target_edge,
    "target_bulk": target_bulk,
    "shape": shape,
}
save_path = f"../data/experimental/pts/edge/{data_id}/pt_e_{shape}.npz"
os.makedirs(os.path.dirname(save_path), exist_ok=True)
save_pts_data(data_edge, save_path, meta_data)

#### Visualize data ####
data = load_pts_data(f"../data/experimental/pts/edge/{data_id}/pt_e_{shape}.npz", perm=(2,1,0))
edge = data["points"][data["label"] == 1]
bulk = data["points"][data["label"] == 0]
overlays = [
    {"data": edge, "label": "Edge", "color": "tab:red"},
    {"data": bulk, "label": "Bulk", "color": "tab:blue"},
]
overlay_points_on_cryoet(cryoet=cryoet, overlays=overlays, axis='x', slice_index=slices, cryo_alpha=0.8, show_title=True)
```
